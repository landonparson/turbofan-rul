Loaded FD001 training data: 100 engines, 20,631 cycles total, 21 sensors. Computed RUL as each engine’s final cycle minus the current cycle





```python
import pandas as pd

cols = ["unit", "cycle", "op1", "op2", "op3"] + [f"s{i}" for i in range(1, 22)]
train = pd.read_csv("../data/raw/train_FD001.txt", sep=r"\s+", header=None, names=cols)
train.head()
```

```python
train["RUL"] = train.groupby("unit")["cycle"].transform("max") - train["cycle"]
train[train["unit"] == 1][["unit", "cycle", "RUL"]].tail()
```
